```python
#Find the Maximum Circular Subarray Sum
#The array is circular, meaning the first element comes after the last element.
#Input

#5 -3 5

#Output

#10

#Explanation:

#5 + 5 = 10

arr = list(map(int, input("Enter an array:").split()))

current_max = arr[0]
maximum_sum = arr[0]

for i in range(1, len(arr)):
  current_max = max(arr[i], current_max+arr[i])
  maximum_sum = max(current_max, maximum_sum)

current_min = arr[0]
minimum_sum = arr[0]

for i in range(1, len(arr)):
  current_min = min(arr[i], current_min + arr[i])
  minimum_sum = min(current_min, minimum_sum)

total = sum(arr)

if maximum_sum < 0:
  print("Maximum Circular Sum:",maximum_sum)
else:
  circular = total - minimum_sum
  print("Maximum Circular Sum:", max(maximum_sum, circular))
```

```python
#Maximum Product Subarray

arr = list(map(int, input("Enter an array:").split()))

max_product = arr[0]
min_product = arr[0]
result = arr[0]

for i in range(1, len(arr)):
  candidates = (arr[i], arr[i]*max_product, arr[i]*min_product)
  max_product = max(candidates)
  min_product = min(candidates)

  result = max(result, max_product)

print("Maximum Product Subarray:", result)
```

```python
#Find Leaders in an Array (Without Extra Space)
#A leader is an element that is greater than all the elements to its right.
#Input
#Enter array elements:
#7 10 4 10 6 5 2
#Output
#Leaders:
#10 6 5 2

arr = list(map(int, input("Enter array elements:").split()))

max_from_right = arr[-1]

for i in range(len(arr)-2,-1,-1):
  if arr[i] > max_from_right:
    max_from_right = arr[i]

print("Leaders:",max_from_right)
```

```python
#Rearrange Array in Wave Form
#Input
#Enter array elements:
#1 2 3 4 5 6
#Output
#Wave Array:
#2 1 4 3 6 5

arr = list(map(int, input("Enter array elements:").split()))

arr.sort()

for i in range(0,len(arr)-1,2):
  arr[i], arr[i+1] = arr[i+1],arr[i]

print("Wave array:",arr)
```

```python
#Longest Mountain in an Array (LeetCode 845)
#A mountain must satisfy:

#Strictly increasing
#Then strictly decreasing

#Input
#2 1 4 7 3 2 5
#Output
#5

arr = list(map(int, input("Enter an array:").split()))
n = len(arr)
maximum = 0

for i in range(1,n-1):
  if arr[i] > arr[i-1] and arr[i] > arr[i+1]:
    left = i
    right = i

    while left > 0 and arr[left] > arr[left-1]:
      left -= 1
    while right < n-1 and arr[right] > arr[right+1]:
      right += 1

    length = right - left + 1
    maximum = max(maximum, length)

print("Longest Mountain Length:",maximum)
```
